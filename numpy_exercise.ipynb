## NUMPY

# numpy multiplication

```python
import numpy as np

lst=[1,2,3,4,5]
array=np.array(lst)
result=array*2
print(result)
```

# numpy division

```python
import numpy as np

lst=[4,8,12,16,20]
array=np.array(lst)
result=array//2
print(result)
```

# numpy addition

```python
import numpy as np

lst=[1,2,3,4,5]
x=[4,8,10,12,16]
array1=np.array(lst)
array2=np.array(x)
result=array1+array2
print(result)
```

# numpy substraction

```python
lst=[1,2,3,4,5]
x=[4,8,10,12,16]
array1=np.array(lst)
array2=np.array(x)
result=array2-array1
print(result)
```

#scalar arithemetic(mathemetic operation on single value)

```python
arr=np.array([1,2,3,4,5])

print(arr + 1)
print(arr - 1)
print(arr * 2)
print(arr / 4)
print(arr ** 3)
```

##vector arithemetic(id array)

```python
array=np.array([1,2,3,4,5,6])
array1=np.array([1.0,2.3,5.1,8.6])
print(np.sqrt(array))
print(np.round(array1))
print(np.floor(array1))
print(np.pi)
```

# slicing

```python
import numpy as np

lst= [1,2,3,4,5,6,7,8,9]
result = np.array(lst)


print("1. result[1:8] - Elements from index 1 to 7:", result[1:8])
print("2. result[1:]  -Elements from index 1 to end:",
result[1:])
print("3. result[:1] - Elements from start to index 0:", result[::-1])
```

# negative slicing

```python
import numpy as np 

arr = np.array([1,2,3,4,5,6,7])

print("4. arr[-3:-1] - from 3rd last to 2nd last element:", arr[-3:-1])
print("5. arr[::3] - every 3rd element:", arr[::3])
```

# slicing 2D array

```python
nested_list = [
	[1,2,3,5,9,8],
	[4,5,6,7,8,9]
]


#---------slicing Row && Column--------#

new_arr = np.array(nested_list)


print(new_arr[0, 0:4])
print(new_arr[0:2, 5])
print(new_arr[0:2, 1:4])
```
